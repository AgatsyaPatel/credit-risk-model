```python
import pandas as pd    # tables and data manipulation
import numpy as np     # math and numbers
import os              # file and folder navigation

os.chdir(r"C:\Users\imagatsya\Downloads\credit-risk-model-main")
print("Working folder:", os.getcwd())
print("Data folder:", os.listdir("data"))
```

```python
columns_we_need = [

    'annual_inc',             
    'emp_length',              
    'home_ownership',          
    'verification_status',    

    # --- LOAN CHARACTERISTICS ---
    # What are the terms of this loan?
    'loan_amnt',               
    'term',                   
    'int_rate',               
    'installment',             
    'grade',                   
    'purpose',                

    # --- CREDIT HISTORY ---
    # What has this person done in the past?
    'fico_range_low',          
    'dti',                     # debt to income — already stretched thin?
    'delinq_2yrs',             
    'mths_since_last_delinq',  # how long ago — recent = still risky
    'pub_rec',                 # bankruptcies — major red flag
    'revol_util',              
    'open_acc',               
    'total_acc',               # length of credit history
    'revol_bal',               # outstanding revolving debt
    'addr_state',              # state economy affects default rates

    # --- TARGET VARIABLE ---
  
    'loan_status',             
]

print(f"We are loading {len(columns_we_need)} columns")
print("Columns:", columns_we_need)
```

```python
print("Loading data in chunks... please wait")
chunks = []

chunk_number = 0

for chunk in pd.read_csv(
    'data/accepted_2007_to_2018Q4.csv',
    usecols=columns_we_need,
    chunksize=50000,
    low_memory=False,
    on_bad_lines='skip'
):
    chunks.append(chunk)
    chunk_number += 1
    print(f"  Loaded chunk {chunk_number} — {chunk_number * 50000:,} rows so far...")

df = pd.concat(chunks, ignore_index=True)
del chunks

print("\n✓ Done!")
print(f"Rows   : {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
```

```python
# See first 5 rows
print("=== FIRST 5 ROWS ===")
display(df.head())

# See data types of each column
print("\n=== COLUMN DATA TYPES ===")
print(df.dtypes)

# See basic statistics for numeric columns
print("\n=== BASIC STATISTICS ===")
display(df.describe())
```

```python
# ── SAVE LOADED DATA ─────────────────────────
df.to_csv("outputs/01_loaded_data.csv", index=False)

print(f"Rows saved: {df.shape[0]:,}")
print(f"Columns   : {df.shape[1]}")
print("✓ Saved to outputs/01_loaded_data.csv")
```
